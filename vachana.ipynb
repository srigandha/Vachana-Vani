# Vachana Vani Local Server

This notebook starts a local HTTP server to serve the Vachana Vani project.

Run the cell below to start the server and open the template in your browser.

```python
import http.server, socketserver, threading, webbrowser

PORT = 8000
Handler = http.server.SimpleHTTPRequestHandler
httpd = socketserver.TCPServer(("", PORT), Handler)
threading.Thread(target=httpd.serve_forever, daemon=True).start()
webbrowser.open(f"http://localhost:{PORT}/Template.html?weekNumber=17&year=2026")

print(f"Server running at http://localhost:{PORT}/")
print("Press Ctrl+C in the terminal to stop the server")
```
